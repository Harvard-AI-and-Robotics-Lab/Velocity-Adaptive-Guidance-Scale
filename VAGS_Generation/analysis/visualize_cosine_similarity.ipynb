```python
import pandas as pd

# Read the input CSV — keep prompt_id as a string to preserve leading zeros
df = pd.read_csv("input.csv", dtype={"prompt_id": str})

# Group by prompt_id and aggregate cos_sim
summary = (
    df.groupby("prompt_id")["cos_sim"]
      .agg(min_cos_sim="min", max_cos_sim="max", avg_cos_sim="mean")
      .reset_index()
)

# Write to a new CSV
summary.to_csv("data4paper/prompt_cos_sim.csv", index=False)

print(summary.head())
```
